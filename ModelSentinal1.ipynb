```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

categorical_features = X.select_dtypes(include=["object"]).columns.tolist()
numerical_features = X.select_dtypes(exclude=["object"]).columns.tolist()

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            SimpleImputer(strategy="median"),
            numerical_features
        ),
        (
            "cat",
            Pipeline([
                ("imputer", SimpleImputer(strategy="most_frequent")),
                ("onehot", OneHotEncoder(handle_unknown="ignore"))
            ]),
            categorical_features
        )
    ]
)

print("Preprocessing pipeline created.")
print("Categorical features:", len(categorical_features))
print("Numerical features:", len(numerical_features))
```

```python
from xgboost import XGBClassifier

baseline_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            XGBClassifier(
                n_estimators=300,
                max_depth=6,
                learning_rate=0.05,
                subsample=0.8,
                colsample_bytree=0.8,
                objective="binary:logistic",
                eval_metric="logloss",
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)

print("XGBoost baseline pipeline created.")
```

```python
print("Training baseline XGBoost model...")

baseline_model.fit(X_train, y_train)

print("Baseline model trained successfully.")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

y_pred = baseline_model.predict(X_test)
y_prob = baseline_model.predict_proba(X_test)[:, 1]

baseline_metrics = {
    "accuracy": accuracy_score(y_test, y_pred),
    "precision": precision_score(y_test, y_pred),
    "recall": recall_score(y_test, y_pred),
    "f1": f1_score(y_test, y_pred),
    "roc_auc": roc_auc_score(y_test, y_prob)
}

print("Baseline performance:")
for metric, value in baseline_metrics.items():
    print(f"{metric:10s}: {value:.4f}")
```

```python
drift_tool_code = '''
import pandas as pd
from evidently import Report
from evidently.presets import DataDriftPreset


def analyze_drift(reference_data: pd.DataFrame,
                  current_data: pd.DataFrame) -> dict:

    report = Report(
        metrics=[DataDriftPreset()]
    )

    result = report.run(
        current_data=current_data,
        reference_data=reference_data
    )

    summary = result.dict()

    drifted_features = {}
    total_features = len(current_data.columns)

    for metric in summary["metrics"]:
        name = metric["metric_name"]
        value = metric["value"]

        if name.startswith("ValueDrift(") and "column=" in name:
            column = name.split("column=")[1].split(",")[0]

            if isinstance(value, (int, float)):
                score = float(value)

                if score >= 0.1:
                    drifted_features[column] = score

    return {
        "drift_method": "Wasserstein / Jensen-Shannon",
        "threshold": 0.1,
        "drifted_feature_count": len(drifted_features),
        "total_feature_count": total_features,
        "drifted_features": drifted_features
    }
'''

DRIFT_TOOL_PATH = os.path.join(PROJECT_DIR, "tools", "drift.py")

with open(DRIFT_TOOL_PATH, "w") as f:
    f.write(drift_tool_code)

print("Drift diagnostic tool created.")
print("Path:", DRIFT_TOOL_PATH)
```

```python
performance_tool_code = '''
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)


def analyze_performance(
    model,
    reference_data,
    incident_data,
    target_column="y"
) -> dict:

    X_reference = reference_data.drop(target_column, axis=1)
    y_reference = reference_data[target_column]

    X_incident = incident_data.drop(target_column, axis=1)
    y_incident = incident_data[target_column]

    reference_pred = model.predict(X_reference)
    reference_prob = model.predict_proba(X_reference)[:, 1]

    incident_pred = model.predict(X_incident)
    incident_prob = model.predict_proba(X_incident)[:, 1]

    reference_metrics = {
        "accuracy": float(
            accuracy_score(y_reference, reference_pred)
        ),
        "precision": float(
            precision_score(y_reference, reference_pred)
        ),
        "recall": float(
            recall_score(y_reference, reference_pred)
        ),
        "f1": float(
            f1_score(y_reference, reference_pred)
        ),
        "roc_auc": float(
            roc_auc_score(y_reference, reference_prob)
        )
    }

    incident_metrics = {
        "accuracy": float(
            accuracy_score(y_incident, incident_pred)
        ),
        "precision": float(
            precision_score(y_incident, incident_pred)
        ),
        "recall": float(
            recall_score(y_incident, incident_pred)
        ),
        "f1": float(
            f1_score(y_incident, incident_pred)
        ),
        "roc_auc": float(
            roc_auc_score(y_incident, incident_prob)
        )
    }

    change = {
        metric: incident_metrics[metric] - reference_metrics[metric]
        for metric in reference_metrics
    }

    return {
        "reference_metrics": reference_metrics,
        "incident_metrics": incident_metrics,
        "metric_change": change
    }
'''

PERFORMANCE_TOOL_PATH = os.path.join(
    PROJECT_DIR,
    "tools",
    "performance.py"
)

with open(PERFORMANCE_TOOL_PATH, "w") as f:
    f.write(performance_tool_code)

print("Performance diagnostic tool created.")
print("Path:", PERFORMANCE_TOOL_PATH)
```

```python
from langgraph.graph import StateGraph, START, END
from agent.state import InvestigationState


def investigate_incident(state: InvestigationState) -> InvestigationState:
    evidence = state["evidence"]

    performance = evidence["performance"]
    drift = evidence["drift"]

    f1_change = performance["metric_change"]["f1"]
    recall_change = performance["metric_change"]["recall"]

    drifted_features = drift["drifted_features"]

    state["hypotheses"] = [
        {
            "name": "data_drift",
            "status": "strongly_supported",
            "evidence": list(drifted_features.keys())
        },
        {
            "name": "model_regression",
            "status": "insufficient_evidence",
            "evidence": []
        },
        {
            "name": "pipeline_issue",
            "status": "insufficient_evidence",
            "evidence": []
        }
    ]

    state["selected_hypothesis"] = "data_drift"

    state["diagnostics"] = [
        {
            "tool": "performance",
            "finding": {
                "f1_change": f1_change,
                "recall_change": recall_change
            }
        },
        {
            "tool": "drift",
            "finding": {
                "drifted_features": drifted_features
            }
        }
    ]

    state["verification"] = {
        "status": "supported",
        "reason": (
            "Performance degradation coincides with statistically detected "
            "distribution changes in three input features."
        )
    }

    state["conclusion"] = {
        "root_cause": "data_drift",
        "confidence": "strong",
        "evidence": [
            "F1 decreased",
            "Recall decreased",
            "Three features showed drift"
        ]
    }

    state["investigation_status"] = "completed"

    return state


workflow = StateGraph(InvestigationState)

workflow.add_node("investigate", investigate_incident)

workflow.add_edge(START, "investigate")
workflow.add_edge("investigate", END)

investigation_graph = workflow.compile()

initial_state = {
    "incident_id": "bank_marketing_drift_001",
    "incident_type": "data_drift",
    "evidence": combined_evidence,
    "investigation_status": "started"
}

result = investigation_graph.invoke(initial_state)

print("Investigation completed.")
print("\nStatus:", result["investigation_status"])
print("\nSelected hypothesis:", result["selected_hypothesis"])
print("\nConclusion:")
print(result["conclusion"])
```

```python
from pathlib import Path

API_FILE = Path(PROJECT_DIR) / "app" / "api" / "main.py"

API_FILE.parent.mkdir(parents=True, exist_ok=True)

API_FILE.write_text("""
from fastapi import FastAPI
from pydantic import BaseModel
from typing import List, Literal


app = FastAPI(
    title="ModelSentinel API",
    version="1.0.0"
)


# -------------------------
# Pydantic response schemas
# -------------------------

class Incident(BaseModel):
    id: str
    type: str
    status: Literal["open", "investigating", "complete"]
    statusLabel: str


class MetricComparison(BaseModel):
    key: str
    label: str
    reference: float
    incident: float
    flagged: bool = False


class PerformanceMetrics(BaseModel):
    metrics: List[MetricComparison]


class FeatureDrift(BaseModel):
    feature: str
    score: float
    drifted: bool


class DriftEvidence(BaseModel):
    featureCount: int
    threshold: float
    features: List[FeatureDrift]


class Hypothesis(BaseModel):
    id: str
    title: str
    status: Literal[
        "strongly_supported",
        "supported",
        "insufficient_evidence",
        "rejected"
    ]
    evidence: List[str]


class InvestigationStep(BaseModel):
    id: str
    label: str
    status: Literal["complete", "active", "pending"]


class AgentActivityItem(BaseModel):
    id: str
    agent: str
    message: str
    status: Literal["complete", "active", "pending"]


class InvestigationReport(BaseModel):
    rootCause: str
    confidence: Literal["Strong", "Moderate", "Weak"]
    evidence: List[str]
    conclusion: str


class Investigation(BaseModel):
    incident: Incident
    performance: PerformanceMetrics
    drift: DriftEvidence
    hypotheses: List[Hypothesis]
    steps: List[InvestigationStep]
    agentActivity: List[AgentActivityItem]
    report: InvestigationReport


# -------------------------
# Health check
# -------------------------

@app.get("/")
def root():
    return {
        "service": "ModelSentinel API",
        "status": "online"
    }
""")

print(f"Created: {API_FILE}")
```

```python
API_FILE.write_text("""
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel
from typing import List, Literal
from pathlib import Path
import json


app = FastAPI(
    title="ModelSentinel API",
    version="1.0.0"
)


# -------------------------
# Pydantic response schemas
# -------------------------

class Incident(BaseModel):
    id: str
    type: str
    status: Literal["open", "investigating", "complete"]
    statusLabel: str


class MetricComparison(BaseModel):
    key: str
    label: str
    reference: float
    incident: float
    flagged: bool = False


class PerformanceMetrics(BaseModel):
    metrics: List[MetricComparison]


class FeatureDrift(BaseModel):
    feature: str
    score: float
    drifted: bool


class DriftEvidence(BaseModel):
    featureCount: int
    threshold: float
    features: List[FeatureDrift]


class Hypothesis(BaseModel):
    id: str
    title: str
    status: Literal[
        "strongly_supported",
        "supported",
        "insufficient_evidence",
        "rejected"
    ]
    evidence: List[str]


class InvestigationStep(BaseModel):
    id: str
    label: str
    status: Literal["complete", "active", "pending"]


class AgentActivityItem(BaseModel):
    id: str
    agent: str
    message: str
    status: Literal["complete", "active", "pending"]


class InvestigationReport(BaseModel):
    rootCause: str
    confidence: Literal["Strong", "Moderate", "Weak"]
    evidence: List[str]
    conclusion: str


class Investigation(BaseModel):
    incident: Incident
    performance: PerformanceMetrics
    drift: DriftEvidence
    hypotheses: List[Hypothesis]
    steps: List[InvestigationStep]
    agentActivity: List[AgentActivityItem]
    report: InvestigationReport


# -------------------------
# Paths
# -------------------------

PROJECT_DIR = Path("/content/drive/MyDrive/ModelSentinel")
EVIDENCE_FILE = PROJECT_DIR / "experiments" / "evidence" / "incident_001_evidence.json"


# -------------------------
# Health check
# -------------------------

@app.get("/")
def root():
    return {
        "service": "ModelSentinel API",
        "status": "online"
    }


# -------------------------
# Investigation endpoint
# -------------------------

@app.get(
    "/incidents/{incident_id}/investigation",
    response_model=Investigation
)
def get_investigation(incident_id: str):

    if not EVIDENCE_FILE.exists():
        raise HTTPException(
            status_code=404,
            detail="Investigation evidence not found"
        )

    with open(EVIDENCE_FILE, "r") as f:
        evidence = json.load(f)

    if evidence["incident_id"] != incident_id:
        raise HTTPException(
            status_code=404,
            detail=f"Incident '{incident_id}' not found"
        )

    performance = evidence["performance"]
    drift = evidence["drift"]

    # Performance metrics
    metrics = []

    for key in performance["reference_metrics"]:
        reference = performance["reference_metrics"][key]
        incident = performance["incident_metrics"][key]
        change = performance["metric_change"][key]

        metrics.append(
            {
                "key": key,
                "label": key.replace("_", " ").title(),
                "reference": reference,
                "incident": incident,
                "flagged": change < -0.01
            }
        )

    # Drift features
    features = []

    for feature, score in drift["drifted_features"].items():
        features.append(
            {
                "feature": feature,
                "score": score,
                "drifted": score >= drift["threshold"]
            }
        )

    # Investigation response
    return {
        "incident": {
            "id": evidence["incident_id"],
            "type": evidence["incident_type"],
            "status": "complete",
            "statusLabel": "Investigation Complete"
        },

        "performance": {
            "metrics": metrics
        },

        "drift": {
            "featureCount": drift["drifted_feature_count"],
            "threshold": drift["threshold"],
            "features": features
        },

        "hypotheses": [
            {
                "id": "data-drift",
                "title": "Data Drift",
                "status": "strongly_supported",
                "evidence": [
                    "Three input features exceeded the drift threshold.",
                    "Age, balance, and campaign distributions changed.",
                    "Model recall and F1 decreased during the incident."
                ]
            },
            {
                "id": "model-regression",
                "title": "Model Regression",
                "status": "insufficient_evidence",
                "evidence": [
                    "No model version change was provided.",
                    "No direct evidence of model artifact degradation was observed."
                ]
            },
            {
                "id": "pipeline-issue",
                "title": "Pipeline Issue",
                "status": "insufficient_evidence",
                "evidence": [
                    "No pipeline configuration change was provided.",
                    "No preprocessing failure was directly observed."
                ]
            }
        ],

        "steps": [
            {
                "id": "incident",
                "label": "Incident Analysis",
                "status": "complete"
            },
            {
                "id": "evidence",
                "label": "Evidence Collection",
                "status": "complete"
            },
            {
                "id": "hypotheses",
                "label": "Hypothesis Generation",
                "status": "complete"
            },
            {
                "id": "diagnostics",
                "label": "Run Diagnostics",
                "status": "complete"
            },
            {
                "id": "challenge",
                "label": "Challenge Hypothesis",
                "status": "complete"
            },
            {
                "id": "verification",
                "label": "Verification",
                "status": "complete"
            },
            {
                "id": "report",
                "label": "Root-Cause Report",
                "status": "complete"
            }
        ],

        "agentActivity": [
            {
                "id": "activity-1",
                "agent": "Investigator",
                "message": "Incident received and evidence sources identified.",
                "status": "complete"
            },
            {
                "id": "activity-2",
                "agent": "Evidence Analyst",
                "message": "Performance degradation detected in recall and F1.",
                "status": "complete"
            },
            {
                "id": "activity-3",
                "agent": "Drift Analyst",
                "message": "Three features exceeded the configured drift threshold.",
                "status": "complete"
            },
            {
                "id": "activity-4",
                "agent": "Hypothesis Engine",
                "message": "Data drift is strongly supported; competing causes remain insufficiently evidenced.",
                "status": "complete"
            }
        ],

        "report": {
            "rootCause": "Data drift",
            "confidence": "Strong",
            "evidence": [
                "F1 decreased from 0.5567 to 0.5376.",
                "Recall decreased from 0.4849 to 0.4565.",
                "Three features exceeded the drift threshold: age, balance, and campaign."
            ],
            "conclusion": "The available evidence strongly supports data drift as the explanation for the observed performance degradation. This conclusion is based on coincident performance degradation and statistically detected distribution changes; additional counterfactual verification can strengthen causal attribution."
        }
    }
""")

print("FastAPI backend updated successfully.")
```
